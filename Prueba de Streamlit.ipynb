Ir poniendo en este **%%writefile app.py** todo el codigo que voy montando

```python
%%writefile app.py
import streamlit as st
import pandas as pd # Se añade para que df pueda ser leído en app.py

st.title("Mi primera app")
st.header("Bienvenidos!")
st.subheader("Me llamo Stephy")
st.write("Estoy muy emocionada de presentar esta aplicación web hecha en python ")
st.code("for i in range(8): print(i)")
st.text("Que increible!")
col1, col2 = st.columns(2)

with col1:
    st.header("Aquí la izquierda")
with col2:
    st.header("Aquí la derecha")
with col1:
  if st.button("Dar Click Aquí!"):
    st.balloons()

df = pd.read_csv("/content/data/filtered_data_updated.csv") # Se mueve aquí para que unique_colors pueda ser accedido.
unique_colors = df['baseColour'].unique()

with col2:
  color = st.selectbox("Elige tu color favorito", unique_colors)
  if st.button("Color favorito es..."):
    st.write(color)
nombre = st.text_input("Ingresa tu nombre")
apellido = st.text_input("Ingresa tu Apellido")
edad = st.slider("Elige tu edad", min_value = 18, max_value = 75)
if edad < 40:
  comentario = "Eres muy joven!"
else:
  comentario = "Eres muy mayor!"
st.write(f"Bienvenido(a)", nombre, apellido, comentario)
st.image("/content/data/birthday-cake-by-Vexels.png", width= 300)
if st.button("sorpresa!"):
  st.image("/content/data/pngegg.png", width= 300)
st.dataframe(df.head())
st.subheader("Dataframe filtrado por tu color preferido")
st.dataframe(df[df['baseColour'] == color])
```
