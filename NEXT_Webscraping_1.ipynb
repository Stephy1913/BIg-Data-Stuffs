<h2 style="color: #008080;">Qué es Web Scraping</h2>

- El **web scraping** es un método utilizado por analistas de datos y desarrolladores para **obtener información de páginas web**.  
- Consiste en **recuperar una página web** y luego **analizar su contenido** para extraer la información deseada.  
- Esta técnica es especialmente útil cuando los datos buscados **no están disponibles a través de APIs**.  
- Los datos extraídos pueden ser **limpiados, analizados o almacenados en bases de datos** para tareas posteriores de análisis de datos.


<h2 style="color: #008080;">Estructura de una Web</h2>

Las tecnologías web fundamentales que conforman la estructura de los sitios que queremos extraer son:

- **HTML**: Es la columna vertebral de casi todos los sitios web. HTML, el lenguaje de marcado principal, es esencial para crear páginas web. Contiene todo el contenido visible en una página.

- **CSS**: Este lenguaje de hojas de estilo trabaja junto a HTML y se encarga del aspecto visual de las páginas web. Controla cómo se presentan los elementos HTML, permitiendo una interfaz organizada y atractiva.

- **JavaScript**: Aporta dinamismo a los sitios web. JavaScript permite crear contenido interactivo y animado. Este lenguaje puede modificar el contenido de una página incluso después de que haya sido cargada, haciendo que las páginas sean más dinámicas y reactivas.

📘 En esta lección, trabajaremos con el **HTML** de los sitios web.

### Hecho 1: HTML está construido con etiquetas

En el ámbito del **web scraping**, comprender **HTML (Lenguaje de Marcado de Hipertexto)** es fundamental.

En su esencia, HTML está compuesto por contenido **envuelto en `<etiquetas>`**. Generalmente contiene el contenido textual que deseamos extraer, marcado mediante estas "etiquetas" entre corchetes angulares.  
Estas etiquetas proporcionan **estructura y significado**, indicando al navegador cómo debe mostrarse el contenido.  

El acrónimo "HTML" significa **HyperText Markup Language (Lenguaje de Marcado de Hipertexto)**.

HTML sigue una estructura en forma de árbol, compuesta por etiquetas padre, etiquetas hijas y etiquetas hermanas. Por ejemplo:

```html
<html>
    <head>
        <title>Título de la página</title>
    </head>
    <body>
        <h1>Mi primer encabezado</h1>
        <p>Mi primer párrafo.</p>
    </body>
</html>

Por ejemplo, considera la etiqueta `<strong>`, que indica un formato en **negrita**.  
Si el texto `"21 de enero"` está encapsulado entre una etiqueta de apertura `<strong>` y su correspondiente etiqueta de cierre `</strong>`, se está señalando el inicio y el final del estilo en negrita.  
Este par de etiquetas le indica al navegador que debe mostrar el texto `"21 de enero"` en **negrita**.


### 🏷️ Tipos de etiquetas HTML

Las etiquetas vienen en varios tipos, cada una diseñada para encapsular contenido específico:

- **Encabezados**:  
  `<h1>`, `<h2>`, `<h3>`, `<h4>`, ...

- **Contenido en línea (frases)**:  
  `<b>`, `<strong>`, `<sub>`, `<i>`, `<a>`, ...

- **Contenido embebido**:  
  `<audio>`, `<img>`, `<video>`, `<iframe>`, ...

- **Datos tabulados**:  
  `<table>`, `<tr>`, `<td>`, `<tbody>`, ...

- **Secciones de página**:  
  `<header>`, `<section>`, `<nav>`, `<article>`, ...

- **Metadatos y scripts**:  
  `<meta>`, `<title>`, `<script>`, `<link>`, ...


### 📘 Hecho 2: Las etiquetas pueden tener atributos

Las etiquetas HTML pueden tener **"atributos"**, que se definen dentro de la etiqueta de apertura.

Observa los siguientes ejemplos:

- `<span class="short-desc">`: Aquí, la etiqueta `<span>` tiene un atributo `class` con el valor `"short-desc"`.
- `<div> Zapas Marca Joma X54 </div>`: Esta etiqueta **no contiene atributos**.
- `<div class="price-item" id="offer"> Zapas Marca Joma X54 </div>`: La etiqueta `div` aquí tiene **dos atributos**:
  - `class` con el valor `"price-item"`
  - `id` con el valor `"offer"`
- `<a class="text-monospace" id="name_132" href="http://www.example.com"> Page Content </a>`:  
  Esta etiqueta `a` contiene los siguientes atributos:
  - `class`: con el valor `"text-monospace"`. Recuerda que **class no es único** en la página.
  - `id`: con el valor `"name_132"`. El `id` está pensado para ser un **identificador único** para una etiqueta.
  - `href`: con el valor `"http://www.example.com"`. El `href` representa un **enlace** a otra sección de la página o a un sitio externo.



### 📌 Puntos clave

- El atributo `id` **debe ser único** para cada etiqueta; dos etiquetas **no deben compartir el mismo id**.
- El atributo `class` **no está pensado para ser único**. Se utiliza comúnmente para **agrupar etiquetas** que comparten un mismo estilo o comportamiento.


Para fines de **web scraping**, **no es crucial entender el significado semántico** de etiquetas como `<span>`, o atributos como `class` o `short-desc`.

Lo verdaderamente importante es **reconocer que las etiquetas pueden tener atributos** y **comprender cómo están estructurados**.  
Cuando extraemos contenido, nuestro objetivo es **identificar las etiquetas correctas** dentro del HTML de una página.


### 🎯 Selección de elementos específicos en Web Scraping

Al adentrarse en el **web scraping**, es fundamental poder **apuntar eficientemente a elementos específicos**.  
Para enfocar el contenido que necesitas extraer, considera **filtrar etiquetas** en función de:

- **Nombre de la etiqueta (Tag Name)**: El tipo principal del elemento (por ejemplo, `<div>`, `<a>`, `<p>`).
- **Clase (Class)**: Un descriptor que agrupa múltiples elementos con características similares.
- **ID**: Un identificador único asignado a un elemento específico.
- **Otros atributos**: Propiedades adicionales como `href`, `title` o `lang` que permiten refinar aún más la selección de los elementos de interés.


## PRINCIPALES FUNCIONES


- `find` → Devuelve **un solo elemento** (el primero que encuentra).
- `find_all` → Devuelve una **lista con todos los elementos** que coinciden.
- `text` → Devuelve **todo el texto** del elemento, incluyendo espacios y saltos de línea (`\n`).
- `get_text` → Devuelve **solo el texto**, de forma limpia.
- `get` → Se utiliza para **obtener atributos**, como los enlaces (`href`).
- `name` → Devuelve el **nombre de la etiqueta** (por ejemplo: `"a"`, `"div"`, `"p"`).

# PRINCIPALES FUNCIONES


`find` y `findAll` (o su equivalente `find_all`) son métodos utilizados para **buscar en el árbol de BeautifulSoup** etiquetas que coincidan con un cierto criterio.

#### 1. **`find`**:
- Devuelve solo **la primera etiqueta** que coincida con un conjunto de criterios.
- Es útil cuando sabes que solo hay una etiqueta relevante o solo te interesa la primera aparición.
- Ejemplo: Si tienes múltiples etiquetas `<p>` en una página y usas `soup.find('p')`, obtendrás solo la **primera** etiqueta `<p>`.


#### 2. **`findAll` o `find_all`**:
- Devuelve una **lista** de todas las etiquetas que coincidan con los criterios dados.
- Es útil cuando quieres capturar **todas las apariciones** de una etiqueta o conjunto de etiquetas.
- Ejemplo: Usar `soup.find_all('p')` te dará una lista con **todas** las etiquetas `<p>` en la página.

### Ejemplo diferencia entre text y get_text()

En BeautifulSoup, tanto .text como .get_text() se usan para obtener el contenido de texto dentro de una etiqueta HTML, pero hay algunas diferencias clave:

1. .text (atributo)
Es una propiedad de BeautifulSoup.
Devuelve el texto de un elemento y sus hijos en forma de cadena.
No permite configuraciones adicionales.
2. .get_text() (método)
Es un método de BeautifulSoup.
Permite configuraciones como:
separator → Especifica qué usar para separar fragmentos de texto dentro de una etiqueta.
strip → Elimina espacios en blanco antes y después del texto.

````python
from bs4 import BeautifulSoup

html = """<div>
            <p>Hola</p>
            <p>Mundo</p>
         </div>"""

soup = BeautifulSoup(html, "html.parser")
div = soup.find("div")

# Usando .text
print(div.text)
# Salida:
# Hola
# Mundo

# Usando .get_text()
print(div.get_text(separator=" | ", strip=True))
# Salida:
# Hola | Mundo
````
